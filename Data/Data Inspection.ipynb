## AI-Assisted Anomaly Detection

```python
from sklearn.ensemble import IsolationForest
```

```python
model_df = eda_df[
    [
        "State",
        "Quarter",
        "Units_Reimbursed",
        "Number_of_Prescriptions",
        "Total_Amount_Reimbursed"
    ]
].copy()
```

```python
model_df = model_df.dropna(
    subset=[
        "Units_Reimbursed",
        "Number_of_Prescriptions",
        "Total_Amount_Reimbursed"
    ]
)
```

```python
model_df.shape
```

```python
model_df["Reimbursement_Per_Prescription"] = (
    model_df["Total_Amount_Reimbursed"] /
    model_df["Number_of_Prescriptions"].replace(0, np.nan)
)
```

```python
model_df = model_df.replace([np.inf, -np.inf], np.nan)

model_df = model_df.dropna(
    subset=["Reimbursement_Per_Prescription"]
)
```

```python
model_df.shape
```

```python
features = [
    "Units_Reimbursed",
    "Number_of_Prescriptions",
    "Total_Amount_Reimbursed",
    "Reimbursement_Per_Prescription"
]

X = model_df[features].copy()
```

```python
X.describe()
```

```python
X_log = np.log1p(X.clip(lower=0))
```

```python
X_log.describe()
```

```python
from sklearn.ensemble import IsolationForest
```

```python
model = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42
)
```

```python
model_df["Anomaly_Flag"] = model.fit_predict(X_log)

model_df["Anomaly_Score"] = -model.score_samples(X_log)
```

```python
financial_risk = financial_anomalies[
    [
        "State",
        "Quarter",
        "Units_Reimbursed",
        "Number_of_Prescriptions",
        "Total_Amount_Reimbursed",
        "Reimbursement_Per_Prescription",
        "Anomaly_Score"
    ]
].sort_values(
    ["Anomaly_Score", "Total_Amount_Reimbursed"],
    ascending=[False, False]
)

financial_risk.head(15)
```

```python
financial_risk["High_RPP_Flag"] = (
    financial_risk["Reimbursement_Per_Prescription"] >
    financial_risk["Reimbursement_Per_Prescription"].quantile(0.95)
)
```

```python
financial_risk["High_Reimbursement_Flag"] = (
    financial_risk["Total_Amount_Reimbursed"] >
    financial_risk["Total_Amount_Reimbursed"].quantile(0.95)
)
```

```python
financial_risk["High_Prescription_Flag"] = (
    financial_risk["Number_of_Prescriptions"] >
    financial_risk["Number_of_Prescriptions"].quantile(0.95)
)
```

```python
financial_risk["Review_Priority_Score"] = (
    financial_risk["High_RPP_Flag"].astype(int)
    + financial_risk["High_Reimbursement_Flag"].astype(int)
    + financial_risk["High_Prescription_Flag"].astype(int)
)
```

```python
financial_risk.sort_values(
    ["Review_Priority_Score", "Anomaly_Score"],
    ascending=[False, False]
)[[
    "State",
    "Quarter",
    "Total_Amount_Reimbursed",
    "Number_of_Prescriptions",
    "Reimbursement_Per_Prescription",
    "Anomaly_Score",
    "High_RPP_Flag",
    "High_Reimbursement_Flag",
    "High_Prescription_Flag",
    "Review_Priority_Score"
]].head(15)
```

```python
financial_risk["Risk_Category"] = np.select(
    [
        financial_risk["Review_Priority_Score"] >= 2,
        financial_risk["Review_Priority_Score"] == 1
    ],
    [
        "High Review Priority",
        "Moderate Review Priority"
    ],
    default="Anomaly Only"
)
```

## Financial Risk Analysis

```python
total_reimbursement = model_df[
    "Total_Amount_Reimbursed"
].sum()

total_reimbursement
```

```python
total_prescriptions = model_df[
    "Number_of_Prescriptions"
].sum()

total_prescriptions
```

```python
average_rpp = (
    total_reimbursement /
    total_prescriptions
)

average_rpp
```

```python
median_rpp = model_df[
    "Reimbursement_Per_Prescription"
].median()

median_rpp
```

```python
financial_baseline = pd.DataFrame({
    "Metric": [
        "Total Reimbursement",
        "Total Prescriptions",
        "Average Reimbursement per Prescription",
        "Median Reimbursement per Prescription"
    ],
    "Value": [
        total_reimbursement,
        total_prescriptions,
        average_rpp,
        median_rpp
    ]
})

financial_baseline
```

```python
financial_baseline
```

```python
state_financial = (
    model_df.groupby("State")
    .agg(
        Total_Reimbursement=("Total_Amount_Reimbursed", "sum"),
        Total_Prescriptions=("Number_of_Prescriptions", "sum"),
        Average_RPP=("Reimbursement_Per_Prescription", "mean")
    )
    .sort_values(
        "Total_Reimbursement",
        ascending=False
    )
)

state_financial.head(15)
```

```python
state_financial["Reimbursement_Share_Percent"] = (
    state_financial["Total_Reimbursement"]
    / total_reimbursement
) * 100
```

```python
state_financial.head(10)
```

```python
state_financial_valid = state_financial[
    state_financial.index != "XX"
].copy()
```

```python
state_financial_valid.head(10)
```

```python
state_risk = state_financial_valid.copy()
```

```python
state_risk["High_Reimbursement_Flag"] = (
    state_risk["Total_Reimbursement"] >
    state_risk["Total_Reimbursement"].quantile(0.90)
)
```

```python
state_risk["High_RPP_Flag"] = (
    state_risk["Average_RPP"] >
    state_risk["Average_RPP"].quantile(0.90)
)
```

```python
state_risk["State_Review_Score"] = (
    state_risk["High_Reimbursement_Flag"].astype(int)
    + state_risk["High_RPP_Flag"].astype(int)
)
```

```python
state_risk.sort_values(
    ["State_Review_Score", "Total_Reimbursement"],
    ascending=[False, False]
).head(15)
```

```python
anomaly_by_state = (
    anomalies.groupby("State")
    .size()
    .reset_index(name="Anomaly_Count")
)
```

```python
anomaly_by_state["Anomaly_Share_Percent"] = (
    anomaly_by_state["Anomaly_Count"]
    / len(anomalies)
) * 100
```

```python
state_risk_combined = state_risk.reset_index().merge(
    anomaly_by_state,
    on="State",
    how="left"
)
```

```python
state_risk_combined["Anomaly_Count"] = (
    state_risk_combined["Anomaly_Count"]
    .fillna(0)
    .astype(int)
)

state_risk_combined["Anomaly_Share_Percent"] = (
    state_risk_combined["Anomaly_Share_Percent"]
    .fillna(0)
)
```

```python
state_risk_combined.sort_values(
    ["State_Review_Score", "Anomaly_Count"],
    ascending=[False, False]
).head(15)
```

```python
state_risk_combined["Combined_Review_Flag"] = (
    (state_risk_combined["State_Review_Score"] >= 1)
    | (state_risk_combined["Anomaly_Count"] >= 5)
)
```

```python
state_risk_combined["Review_Category"] = np.select(
    [
        state_risk_combined["State_Review_Score"] >= 2,
        (
            (state_risk_combined["State_Review_Score"] == 1)
            & (state_risk_combined["Anomaly_Count"] >= 5)
        ),
        state_risk_combined["Anomaly_Count"] >= 5
    ],
    [
        "High Financial Review",
        "Financial + Anomaly Review",
        "Anomaly Review"
    ],
    default="Routine Monitoring"
)
```

```python
final_state_review = state_risk_combined.sort_values(
    ["Combined_Review_Flag", "State_Review_Score", "Anomaly_Count"],
    ascending=[False, False, False]
)
```

```python
final_state_review[
    [
        "State",
        "Total_Reimbursement",
        "Reimbursement_Share_Percent",
        "Average_RPP",
        "Anomaly_Count",
        "Anomaly_Share_Percent",
        "State_Review_Score",
        "Review_Category"
    ]
].head(20)
```

```python
final_state_review.to_csv(
    "State Financial Risk Summary.csv",
    index=False
)
```
